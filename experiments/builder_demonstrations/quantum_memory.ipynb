## Demonstrating the Builder module
### Simulation of a quantum memory experiment


```python
from builder.experiments._quantum_memory import QuantumMemory
from dotg.decoders import MinimumWeightPerfectMatching
from dotg.noise import DepolarizingNoise
import numpy as np

qmem = QuantumMemory((3, 3), "X", num_qubits=5)
circuit = qmem.run(3)
circuit
```

```python
logical_error_values = []
for p in np.logspace(-3, -2, 10):
    noise_model = DepolarizingNoise(physical_error=p)
    noisy_circuit = noise_model.permute_circuit(circuit=circuit)
    decoder = MinimumWeightPerfectMatching(circuit=noisy_circuit)
    logical_error_values.append(decoder.logical_error(1e6))
```

```python
qmem.draw().show()
```
