```python
from builder.lattice_surgery._experiment import LatticeSurgeryExperiment
from builder.utilities.grids import SquareGrid
from builder.patches import RotatedSurfaceCode

grid = SquareGrid(4, 4)
code = RotatedSurfaceCode((3, 3), grid, (1, 1))
quantum_memory = LatticeSurgeryExperiment(qubit_grid=grid)
quantum_memory.reset_qubits(qubits=code.data_qubits, basis="RZ")
quantum_memory.reset_qubits(qubits=code.x_stabilizers, basis="RX")
quantum_memory.reset_qubits(qubits=code.z_stabilizers, basis="RZ")
quantum_memory.tick()
quantum_memory._depth_4_syndrome_extraction(patches=[code])
quantum_memory.measure_qubits(qubits=code.x_stabilizers, basis="MX")
quantum_memory.measure_qubits(qubits=code.z_stabilizers, basis="MZ")
quantum_memory.detector_batch(
    qubits=code.x_stabilizers, new_measurements_deterministic=False
)
quantum_memory.detector_batch(
    qubits=code.z_stabilizers, new_measurements_deterministic=True
)
for _ in range(3):
    quantum_memory.syndrome_extraction_with_detectors(patches=[code])

quantum_memory.measure_qubits(qubits=code.data_qubits, basis="MZ")
quantum_memory._increment_measurement_record(batch=code.data_qubits)
quantum_memory.detector_batch_with_data_qubits(
    stabilizers=code.z_stabilizers,
    data_qubit_member_check=code.data_qubits,
    final_round_detector=True,
)
quantum_memory.observable(
    observable_index=0,
    targets=[quantum_memory.measurement_record[q] for q in code.bottom_boundary_data],
)
```
